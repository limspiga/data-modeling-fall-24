# Clasificación de Modelos Generativos

Un modelo generativo es un tipo de modelo de aprendizaje automático diseñado para aprender
- la **distribución de datos** de un conjunto de entrenamiento y
- generar datos similares a los originales.

Su objetivo principal es crear contenido nuevo y realista, manteniendo las características estadísticas y estructurales de los datos originales.

# SentenceTransformer
Un **SentenceTransformer** es una herramienta de Python que convierte frases o documentos en datos numéricos llamados embeddings, que representan su significado. Esto es útil para tareas como buscar textos similares, clasificar información o agrupar temas relacionados.

### ¿Cómo funciona?  
Usa modelos de inteligencia artificial preentrenados, para entender el contexto y significado completo de una oración, en lugar de analizar palabra por palabra.

### Ventajas:  
- **Alta precisión**: Captura el significado completo del texto.  
- **Rápido y eficiente**: Ideal para grandes volúmenes de datos.  
- **Flexible**: Compatible con tareas comunes de análisis de texto y fácil de integrar en proyectos.  

Es una herramienta clave para trabajar con texto de manera inteligente y automatizada.

# Ejercicio 8.01 Sistema de recomendación
Usa **SentenceTransformer** para hacer una recomendación que identifique el auto más relevante para un cliente, basado en una descripción de sus necesidades en lenguaje natural. El sistema debe procesar consultas personalizadas, comparar semánticamente las descripciones con un inventario de autos y ofrecer recomendaciones precisas.

```python
from sentence_transformers import SentenceTransformer

# Carga un modelo preentrenado
model = SentenceTransformer('all-MiniLM-L6-v2')

# Inventario de autos con descripciones
autos = [
    "SUV familiar con amplio espacio interior y excelentes características de seguridad.",
    "Sedán deportivo con motor turbo y diseño elegante.",
    "Camioneta todo terreno, ideal para aventuras y uso en terrenos difíciles.",
    "Auto compacto económico con alta eficiencia de combustible, ideal para la ciudad."
]

# Consulta del cliente
consulta_cliente = "Busco un SUV familiar, espacioso y seguro para viajes largos."

# Genera embeddings para la consulta y el inventario
embedding_consulta = model.encode(consulta_cliente)
embedding_autos = model.encode(autos)

# Calcula la similitud entre la consulta y cada auto del inventario
from sklearn.metrics.pairwise import cosine_similarity
similitudes = cosine_similarity([embedding_consulta], embedding_autos)[0]

# Encuentra el auto más relevante
indice_mejor_auto = similitudes.argmax()
mejor_auto = autos[indice_mejor_auto]

# Resultado
print(f"El auto más relevante para la consulta del cliente es:\n{mejor_auto}")
```

```python
# Función para realizar una consulta basada en similitud de embeddings
def buscar_productos_similares(descripcion_consulta, top_n=3):
    # Generar el embedding de la consulta
    embedding_consulta = model.encode(descripcion_consulta).reshape(1, -1)

    %sql resultados << SELECT json_agg(products) FROM products;

#     # Calcular similitud de coseno entre el embedding de consulta y cada embedding en la base de datos
    descripciones = []
    embeddings = []
    products = resultados[0][0]
    for product in products:
      print(product)

      cat = json.dumps(product)
      descripciones.append(cat)
      embedding = model.encode(cat, normalize_embeddings=True).tolist()
      embeddings.append(np.array(embedding))

#     # Convertir a array de numpy para calcular similitudes
    embeddings = np.array(embeddings)
    similitudes = cosine_similarity(embedding_consulta, embeddings).flatten()

    # Obtener los índices de los top_n productos más similares
    indices_similares = np.argsort(similitudes)[::-1][:top_n]

    # Retornar las descripciones de los productos más similares
    productos_similares = [(descripciones[i], similitudes[i]) for i in indices_similares]
    return productos_similares
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer

local_path = "./models/Llama-3.2-1B"
model_id = "meta-llama/Llama-3.2-1B"

# Descarga el modelo y el tokenizador
model = AutoModelForCausalLM.from_pretrained(model_id, cache_dir=local_path)
tokenizer = AutoTokenizer.from_pretrained(model_id, cache_dir=local_path)
```

```python
from transformers import pipeline
pipe = pipeline("text-generation", model=model_id)
```

```python
text_generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=128
)
```

```python
def get_response(prompt):
  response = text_generator(prompt)
  return response[0]['generated_text']

prompt="Qué es la mercadotecnía?"
get_response(prompt)
```
